#### Checking price calculation

```python
price_cols = ['price', 'price_quote_total_price', 'price_quote_price_per_night', 'price_quote_checkin_date', 'price_quote_checkout_date', 'price_quote_raw']
price_cols_ext = price_cols + ['minimum_nights', 'estimated_revenue_l365d']
```

```python
# create a new column 'number_of_nights' by calculating the difference between 'price_quote_checkout_date' and 'price_quote_checkin_date'
listings['number_of_nights'] = pd.to_datetime(listings['price_quote_checkout_date']) - pd.to_datetime(listings['price_quote_checkin_date'])
listings['number_of_nights']
```

```python
# checking if price_quote_price_per_night is equal to price_quote_total_price / number_of_nights
listings['test_price_quote_price_per_night'] = round(listings['price_quote_total_price'] / listings['number_of_nights'].dt.days, 2)
listings['flag_price_quote_price_per_night'] = listings['price'] == listings['test_price_quote_price_per_night']
listings['diff_price_quote_price_per_night'] = abs(listings['price'] - listings['test_price_quote_price_per_night'])
listings[(listings['price'].notna()) & (listings['price_quote_total_price'].notna())
         & (listings['flag_price_quote_price_per_night'] == False)][['price', 'price_quote_total_price', 'test_price_quote_price_per_night', 'flag_price_quote_price_per_night', 'diff_price_quote_price_per_night']]
```

price equals price_quote_total_price / nights for 99.4% of listings with a quote (8,050 rows); the remaining 48 differ by exactly 0.01, consistent with rounding. price_quote_price_per_night is identical to price wherever both exist (107 rows have price only). Core keeps price and the quote check-in/check-out dates; price_quote_price_per_night, price_quote_total_price and price_quote_raw stay in staging only.

**`estimated_occupancy_l365d` / `estimated_revenue_l365d`.** Not documented in the Inside Airbnb
data dictionary. In the records inspected, `estimated_revenue_l365d` equals
`estimated_occupancy_l365d` × `price`, and occupancy appears to follow recent review activity
(0 when there are no recent reviews). Both are therefore treated as modelled estimates, not
observed bookings. They are kept in core with this caveat and are not used as independent
evidence when analysing price effects (circularity risk).
