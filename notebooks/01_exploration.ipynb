#### Import listings.csv.gz

```python
listings = pd.read_csv('../data/raw/listings.csv.gz', compression='gzip')
print(f"Rows: {listings.shape[0]}, Columns: {listings.shape[1]}")
print(f"Total elements: {listings.size}")
```

#### Null Analysis

```python
listings.isna().sum().sort_values(ascending=False).head(20)
```

```python
listings.isna().sum().sort_values(ascending=False)
```

```python
listings_perc_null = round(listings.isna().sum()/listings.shape[0], 4).sort_values(ascending=False)
listings_perc_null
```

#### Checking id

```python
print(f"Rows: {listings.shape[0]}")
print(f"Unique IDs: {listings['id'].nunique()}")
print(f"Missing IDs: {listings['id'].isna().sum()}")
```

#### Checking price

```python
listings['price'].head(10) # checking 10 records not null, need to remove the $ and , to convert to float
```

```python
listings['price'] =listings['price'].str.strip('$').str.replace(',', '').astype(float)
```

```python
listings['price'].info()
```

```python
print(f"Mean: {listings['price'].mean()}")
print(f"Median: {listings['price'].median()}")
print(f"Max: {listings['price'].max()}")
print(f"Min: {listings['price'].min()}")
```
